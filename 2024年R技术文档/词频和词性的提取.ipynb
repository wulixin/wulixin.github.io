```python
import jieba 
```

```python
txt_seg=jieba.lcut(txt)
```

```python
txt_seg[:10]
```

```python
from collections import Counter
```

```python
word_freq=Counter(txt_seg)
```

```python
word_freq.most_common()
```

```python
words_seg=pseg.cut(txt_seg)
```

```python
for w,f in words_seg.items()
     print(w,"出现了",f,'次')
```
